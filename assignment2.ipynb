1-Write a function that takes a list of strings and a number K. The function should
return a new list containing only the strings that have a length greater than K.

```python
def filterTheList(textList,k):
    newList=[]
    for t in textList:
        if len(t) > k :
            newList.append(t)
    return newList

list=["apple","pie","banana","tea","orange"]
k=3
print(filterTheList(list,k))
    
```

2-Write a function named area() that calculates the area of either a "circle" or a
"rectangle". The function should take the shape name as the first argument,
followed by the necessary dimensions (radius for circle; length and width for
rectangle).

```python
import math
def area():
    area=0
    shape=input("enter the shape name")
    shape=shape.lower()
    if shape=="rectangle":
        width=float(input("enter the width of the rectangle"))
        length=float(input("enter the length of the rectangle"))
        area=width*length
        return area
    elif shape=="circle":
        radius=float(input("enter the radius of the circle"))
        area=math.pi * pow(radius,2)
        return area
    else :
        print(" invalid shape")
        return area

print(area())
        
```

4- Write a program to find the sum of all numbers in a list. ensure
the program ignores any elements in the list that are not numbers (like strings or
booleans).

```python
def sum(list):
    sum=0
    for element in list:
        if isinstance(element,(float,int)) and not isinstance(element,(bool,str)):
            sum=sum+element
    return sum
print(sum([10,"apple", 20, True, 5]))
```

5-Write a Python class Restaurant with attributes like menu_items, book_table, and customer_orders, and methods like add_item_to_menu, book_tables, and customer_order. (Note: Use dictionaries and lists to store the data.)

```python
class Restaurant:
    def __init__(self,menu_items,available_tables,book_table):
        self.menu_items=menu_items
        self.available_tables=available_tables
        self.book_table=book_table
        self.customer_orders=[]
    
    def add_item(self):
        item=input("enter the item")
        if item not in self.menu_items:
            self.menu_items.append(item)
        else:
            print("item already exist")
            
    def book_a_table(self,table_number):
        if table_number in self.available_tables:
            if table_number not in self.book_table:
                print (" reservation completed ")
                self.book_table.append(table_number)
            else :
                print(" table not available right now")
        else :
            print("there is no table with that number")
            
    def take_customer_order(self):
        while True:
            order=input("enter item or exit").lower()
            if order =="exit":
                break
            else:
                if order not in self.menu_items:
                    print("item not available")
                else:
                    self.customer_orders.append(order)
        print("your order will be ready soon")
        return self.customer_orders
               
            
            


restaurant=Restaurant(["chicken","pasta","rice"],[1,2,3,4,5,6,7],[2,4,6])
restaurant.add_item()
restaurant.menu_items

restaurant.book_a_table(2)
restaurant.book_table

restaurant.take_customer_order()
```
