##### Create a 6x4 matrix of random integers. Identify which rows have a sum that is an even number. Print the indices of these rows

```python
array=np.random.randint(1,100,size=(6,4))
indices=[]
for i in range(len(array)):
    if array[i].sum() % 2==0:
        indices.append(i)
print(array)
print(indices)
```

##### Create a 1D array of 30 random integers between 1 and 50. Replace all numbers that are multiples of 3 or 5 with the value -1

```python
arr=np.random.randint(1,50,30)
print(arr)
for i in range(len(arr)):
    if arr[i] % 3 ==0 or arr[i]%5==0:
        arr[i]=-1
print(arr)
```

##### Write a NumPy program to create a 3x3 array with random values. For each row,calculate its mean (average) and subtract that mean from every element in that specific row.

```python
array=np.random.randint(1,100,size=(3,3))
print(array)
for i in range(len(array)):
    mean=array[i].mean()
    # to ignore float numbers
    mean=np.ceil(mean)
    print("for",i,"row mean=",mean)
    for element in range(len(array[i])):
        array[i][element]=array[i][element]-mean
print(array)
```
